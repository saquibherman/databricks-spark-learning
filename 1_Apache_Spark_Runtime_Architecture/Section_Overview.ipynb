# 1. Apache Spark Runtime Architecture
### Duration - 2 Minutes

This section introduces Apache Spark and its distributed runtime architecture, then explores how that architecture appears in a live Databricks cluster.

## Learning Objectives

**By the end of this section, you will be able to:**

- Describe what Apache Spark is and where it came from
- Identify the roles of the Driver, Workers, and Executors in Spark applications
- Explain how the `SparkSession` serves as the unified entry point for Spark applications
- Use the Spark UI to monitor application progress, DAG visualizations, and resource allocation
