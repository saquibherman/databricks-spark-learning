# 1.1 Lecture - What is Apache Spark
### Duration - 10 Minutes

>  *A unified large-scale data processing and analytics framework*

This lecture introduces Apache Spark: what it is, where it came from, and what makes up the Spark 4 component stack.

## Learning Objectives

**By the end of this lecture, you will be able to:**

- Describe Apache Spark and the workloads it supports
- Summarize the history of the Spark project and its relationship to Databricks

<div style="font-size: 1.2em; font-weight: bold; color: #1B5162; margin: 12px 0 4px 0">The Libraries</div>
<ul style="font-size: 1.1em; line-height: 1.45; color: #333; margin: 0 0 10px 0; padding-left: 24px">
<li style="margin: 4px 0"><strong>Structured Streaming</strong> processes unbounded data using the same DataFrame operations you use for batch, so a streaming pipeline looks almost identical to a batch one.</li>
<li style="margin: 4px 0"><strong>MLlib</strong> provides distributed implementations of common machine learning algorithms and feature transformers.</li>
<li style="margin: 4px 0"><strong>Spark SQL</strong> exposes the engine through standard SQL, and shares the optimizer and execution engine with the DataFrame API.</li>
<li style="margin: 4px 0"><strong>pandas API on Spark</strong> offers a pandas-compatible interface for Python users, executing pandas-style code at cluster scale.</li>
<li style="margin: 4px 0"><strong>GraphX</strong> supports graph algorithms; in practice most graph work in Python uses the GraphFrames package built on DataFrames.</li>
</ul>
<div style="font-size: 1.2em; font-weight: bold; color: #1B5162; margin: 14px 0 4px 0">The Foundation Layers</div>
<ul style="font-size: 1.1em; line-height: 1.45; color: #333; margin: 0 0 10px 0; padding-left: 24px">
<li style="margin: 4px 0">The <strong>DataFrame API</strong> is the primary programming interface in modern Spark. It is declarative: you describe the result you want, and the Catalyst optimizer determines how to compute it.</li>
<li style="margin: 4px 0"><strong>Spark Core</strong> implements scheduling, memory management, fault recovery, and the low-level RDD (Resilient Distributed Dataset) abstraction. In Spark 4 you rarely program against RDDs directly, and the RDD API is not available over Spark Connect, but everything still executes on this layer.</li>
<li style="margin: 4px 0">The R language binding (SparkR) is deprecated as of Spark 4.0; R users are directed to the community-supported sparklyr package.</li>
</ul>

## D. What is New in Spark 4

This course is based on __Apache Spark 4__, the version that ships with __Databricks Runtime 17.x__. The table below summarizes the headline changes you will encounter in this course.

| Feature | What it Means |
|---------|---------------|
| ANSI SQL mode by default | Invalid casts and arithmetic overflows now raise errors instead of silently returning `NULL`. Use `try_cast()` for permissive conversions. |
| Spark Connect | A client-server architecture that decouples your application from the cluster using a thin client and the gRPC protocol. |
| `VARIANT` data type | A native type for semi-structured data (such as JSON) that is faster and more flexible than storing JSON as strings. |
| Native plotting | Plot directly from a PySpark DataFrame with `df.plot` (backed by plotly), without converting to pandas first. |
| SQL scripting | Procedural SQL with `BEGIN ... END` blocks, variables, conditionals, and loops. The course setup notebooks use this. |
| Arrow-optimized Python UDFs | Python user defined functions can use Apache Arrow for serialization, reducing the Python-JVM overhead. |

You will meet ANSI mode and `try_cast()` in the ETL demo, native plotting in Section 4, and SQL scripting in the course setup notebooks.
