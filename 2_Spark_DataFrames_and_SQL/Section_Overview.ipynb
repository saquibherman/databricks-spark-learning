# 2. Spark DataFrames and SQL
### Duration - 2 Minutes

This section introduces the DataFrame, Spark's structured data abstraction, explains how Spark optimizes and lazily evaluates DataFrame operations, and shows how to read and write data in a variety of formats.

## Learning Objectives

**By the end of this section, you will be able to:**

- Describe the DataFrame abstraction, including schema enforcement and optimized execution
- Explain transformations, actions, and lazy evaluation
- Read from and write to various formats (CSV, JSON, Parquet, tables), handling schema inference and explicit schemas
- Query DataFrames using Spark SQL
