# 4. ETL with the DataFrame API
### Duration - 2 Minutes

This section applies everything so far: a lecture on the core transformation toolkit, a demo performing end-to-end ETL over flight data, and a hands-on lab analyzing transaction data.

## Learning Objectives

**By the end of this section, you will be able to:**

- Apply DataFrame transformation methods and map them to their SQL equivalents
- Reference columns correctly and use Column object methods and built-in functions
- Handle missing and invalid values, including Spark 4 ANSI mode behavior
- Create and apply user defined functions, including vectorized pandas UDFs
- Build and save an end-to-end ETL pipeline
