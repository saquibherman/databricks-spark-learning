# Introduction to Apache Spark
### Duration - 2 Minutes

This beginner-friendly course covers the fundamentals of Apache Spark 4 for large-scale data processing. You will explore Spark's distributed architecture, learn the DataFrame API, and read, write, and process data using Python on the Databricks Data Intelligence Platform. Through hands-on demonstrations and a lab, you will build the skills needed to execute Spark transformations and actions efficiently.

## Terminal Learning Objective

**Process and analyze data at scale using the Apache Spark 4 DataFrame API on Databricks**

By the end of this course, you will be able to:

- Describe the distributed architecture of Apache Spark and the roles of the Driver, Workers, and Executors
- Explain core distributed systems concepts including shared-nothing architecture, partitioning, and the shuffle
- Use the DataFrame API to express transformations and actions, and explain lazy evaluation
- Read data into DataFrames and write DataFrames to files and Unity Catalog tables
- Apply common ETL operations including filtering, type conversion, derived columns, and user defined functions
- Monitor Spark applications using the Spark UI

## Prerequisites

You should meet the following prerequisites before starting this course:

- Basic programming knowledge
- Familiarity with Python
- Basic understanding of SQL queries (`SELECT`, `JOIN`, `GROUP BY`)
- Familiarity with data processing concepts
- No prior Spark or Databricks experience required
